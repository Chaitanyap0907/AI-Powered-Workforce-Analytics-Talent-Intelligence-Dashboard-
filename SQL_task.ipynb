## Basic SQL Operations with SQLite

First, let's import the `sqlite3` library and create an in-memory database. This database will only exist for the duration of this Colab session.

```python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

print("Connected to in-memory SQLite database.")
```

### 1. Create Table

We will create a simple `Products` table with columns for `product_id`, `name`, `category`, and `price`.

```python
create_table_sql = """
CREATE TABLE Products (
    product_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    category TEXT,
    price REAL NOT NULL
);
"""

cursor.execute(create_table_sql)
conn.commit()
print("Table 'Products' created successfully.")
```

### 2. Insert Data

Now, let's insert some random data into our `Products` table.

```python
insert_data_sql = """
INSERT INTO Products (name, category, price) VALUES
    ('Laptop', 'Electronics', 1200.00),
    ('Mouse', 'Electronics', 25.50),
    ('Keyboard', 'Electronics', 75.00),
    ('Desk Chair', 'Furniture', 150.00),
    ('Table Lamp', 'Furniture', 40.00),
    ('Notebook', 'Stationery', 10.00),
    ('Pen Set', 'Stationery', 15.00),
    ('Monitor', 'Electronics', 300.00),
    ('Bookshelf', 'Furniture', 90.00),
    ('Backpack', 'Accessories', 60.00);
"""

cursor.execute(insert_data_sql)
conn.commit()
print("10 records inserted into 'Products' table.")
```

### 3. Basic Queries

Let's perform some basic SQL queries to retrieve and manipulate the data.

#### Select all data

```python
select_all_sql = "SELECT * FROM Products;"
cursor.execute(select_all_sql)

# Fetch all results
results = cursor.fetchall()

# Get column names for DataFrame
column_names = [description[0] for description in cursor.description]

df_all = pd.DataFrame(results, columns=column_names)
display(df_all)
```

#### Select products from a specific category (e.g., 'Electronics')

```python
select_category_sql = "SELECT name, price FROM Products WHERE category = 'Electronics';"
cursor.execute(select_category_sql)
results_electronics = cursor.fetchall()
column_names_electronics = [description[0] for description in cursor.description]
df_electronics = pd.DataFrame(results_electronics, columns=column_names_electronics)
display(df_electronics)
```

#### Select products with price greater than a certain value (e.g., 100.00)

```python
select_price_sql = "SELECT name, category, price FROM Products WHERE price > 100.00;"
cursor.execute(select_price_sql)
results_price = cursor.fetchall()
column_names_price = [description[0] for description in cursor.description]
df_price = pd.DataFrame(results_price, columns=column_names_price)
display(df_price)
```

#### Order products by price in descending order

```python
order_by_price_sql = "SELECT name, price FROM Products ORDER BY price DESC;"
cursor.execute(order_by_price_sql)
results_ordered = cursor.fetchall()
column_names_ordered = [description[0] for description in cursor.description]
df_ordered = pd.DataFrame(results_ordered, columns=column_names_ordered)
display(df_ordered)
```

#### Calculate the average price of products by category

```python
avg_price_sql = "SELECT category, AVG(price) as average_price FROM Products GROUP BY category;"
cursor.execute(avg_price_sql)
results_avg_price = cursor.fetchall()
column_names_avg_price = [description[0] for description in cursor.description]
df_avg_price = pd.DataFrame(results_avg_price, columns=column_names_avg_price)
display(df_avg_price)
```

#### Update a product's price

```python
update_price_sql = "UPDATE Products SET price = 1250.00 WHERE name = 'Laptop';"
cursor.execute(update_price_sql)
conn.commit()
print("Price of Laptop updated.")

# Verify the update
verify_update_sql = "SELECT name, price FROM Products WHERE name = 'Laptop';"
cursor.execute(verify_update_sql)
print(cursor.fetchone())
```

#### Delete a product

```python
delete_product_sql = "DELETE FROM Products WHERE name = 'Pen Set';"
cursor.execute(delete_product_sql)
conn.commit()
print("Product 'Pen Set' deleted.")

# Verify the deletion (should not find 'Pen Set')
verify_delete_sql = "SELECT * FROM Products WHERE name = 'Pen Set';"
cursor.execute(verify_delete_sql)
print(f"Found 'Pen Set' after deletion: {cursor.fetchone()}")
```

### Closing the connection

It's good practice to close the database connection when you are done.

```python
conn.close()
print("Database connection closed.")
```
